# Random Walk Simulator


## 1D Random Walk

```python
import numpy as np
import matplotlib.pyplot as plt

steps=2000
origin=0

step_set=np.array([-1,1])
random_steps=np.random.choice(step_set,size=steps)

path=np.concatenate([[origin],np.cumsum(random_steps)])

plt.figure(figsize=(10,4))
plt.plot(path,color='#2ca02c',linewidth=1)
plt.title("1D Random walk simulator")
plt.xlabel("Steps")
plt.ylabel("Positions")
plt.grid(True,alpha=0.3)
plt.plot()
```

## 2D Random Walk

```python
import numpy as np
import matplotlib.pyplot as plt

n=10000
moves=np.array([(0,1),(0,-1),(1,0),(-1,0)])

random_indices=np.random.choice(len(moves),size=n)
steps=moves[random_indices]

coords=np.cumsum(steps,axis=0)
x,y=coords[:,0],coords[:,1]

plt.figure(figsize=(8,8))
plt.plot(x,y,lw=0.5,color='blue')
plt.scatter(0,0,color='red',label='start')
plt.scatter(x[-1],y[-1],color='black',label='End')
plt.title(f"2D Random Walk({n} Steps)")
plt.legend()
plt.axis('equal')
plt.show()
```
